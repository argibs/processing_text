## Set your Working Directory and Initialize the .txt file you need

```python
import os
print(os.getcwd()) # This shows you your current working directory

# We want to set the working directory to the folder where our text files are saved. Change this to your own file path.
wd = r"C:\Users\ashto\OneDrive\Desktop\capstone_append\Capstone_Interview"
# change the working directory to the path above (string variabe wd)
os.chdir(wd)

# check to make sure it worked! 
print(os.getcwd())

# Initialize the text file we want to edit
input_file_path = r"C:\Users\ashto\OneDrive\Desktop\capstone_append\Capstone_Interview\GI13_transcript.txt"
output_file_path = r"C:\Users\ashto\OneDrive\Desktop\capstone_append\Capstone_Interview\GI13_cleaned.txt"
```

## Open the .txt file and read line-by-line, look for punctuation, and save to new _cleaned file

```python
import re # regex module

with open(input_file_path, "r", encoding="utf-8") as infile, \
     open(output_file_path, "w", encoding="utf-8") as outfile:

    for line in infile: # reading the file line-by-line
        line = line.rstrip("\r\n")

        # uses regular expressions (regex) to search for punctuation. Add additional punctuation marks between the [] brackets as needed
        edited_line = re.sub(r"([.?])(?!\n)", r"\1\n", line)  # (?!\n) is regex meaning "not followed by a newline"

        # write the edited line to the output file and strip trailing newlines
        outfile.write(edited_line.rstrip("\n") + "\n")

print(f"Processed file saved to: {output_file_path}") # to verify that it worked
```
